```python
# 6. Check whether a year is a leap year
year = int(input("Enter a year: "))
if year%4==0:
  print(year, "is a leap year")
else:
  print(year, "is not a leap year")
```

```python
# 7. Create a basic calculator
print("=== Basic Calculator ===")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")

choice = input("Choose an operation (1-4): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == "1":
    print("Result:", num1 + num2)

elif choice == "2":
    print("Result:", num1 - num2)

elif choice == "3":
    print("Result:", num1 * num2)

elif choice == "4":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Cannot divide by zero!")

else:
    print("Invalid choice!")
```

```python
# 8. Create a grade calculator
name = input("Enter your name: ")
a = int(input("Enter 1 number: "))
b = int(input("Enter 2 number: "))
c = int(input("Enter 3 number: "))
d = int(input("Enter 4 number: "))
e = int(input("Enter 5 number: "))
total = a+b+c+d+e
perc = (total/500)*100
if perc>=90:
  print("Grade A")
elif perc>=80:
  print("Grade B")
elif perc>=70:
  print("Grade C")
elif perc>=60:
  print("Grade D")

print("Name: ", name)
print("Total:", total)
print("Precentage", perc)

if 60>perc:
  print("Pass")
else:
  print("Fail")
```

## Day 3 Loops & Iteration


```python
# 1. Print numbers 1 - 100
for i in range(1,100+1):
  print(i, end = " ")
```

```python
# 2. Print even numbers 1 -100
for i in range(1,100+1):
  if i%2==0:
    print(i, end=" ")
```

```python
# 3. Print multiplication tables
n = int(input("Enter a number: "))
for i in range(1,10+1):
  print(n, "X", i, "=", n*i)
```

```python
# 4. Calculate sum from 1-N
n = int(input("Enter the number: "))
Total = 0
for i in range(1,n+1):
  Total += i
print(Total)
```

```python
# 5. Calculate Factorial
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n+1):
  fact *= i
print("Factorial = ", fact)
```

```python
# 6. Reverse a number
n = 12345
rev = 0
while n > 0:
  r = n%10
  rev = rev*10+r
  n = n // 10
print(rev)
```

```python
# 7. Count digit
n = 12345
count = 0
while n > 0:
  count += 1
  n = n//10
print(count)
```

```python
# 8. Check palindrom
n = 121
temp = n
rev = 0

while n>0:
  r = n%10
  rev = rev*10+r
  n = n//10

if temp == rev:
  print("Palindrome")
else:
  print("Not a palindrome")
```

```python
# 9. Check prime number
n = 7
count = 0

if n > 1:
    for i in range(1, n + 1):
        if n % i == 0:
            count += 1

if count == 2:
    print("Prime Number")
else:
    print("Not Prime Number")
```

```python
# 10. Fibonacci series

n = int(input("Enter the number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

## Day 4 String

```python
# 1. Reverse a string
s = "Ved Ingole"
s_new = s[::-1]
print(s_new)
```

```python
# 2. Count vowels
s = input("Enter a string: ")

count = 0

for ch in s.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels:", count)
```

```python
# 3. Count consonants
s = input("Enter a string: ")

count = 0

for ch in s.lower():
    if ch.isalpha() and ch not in "aeiou":
        count += 1

print("Number of vowels:", count)
```
