# Python Roadmap

## Day 1 Programming Fundamentals

```python
# 1. Print your name, age and city.
name = input("Enter your name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")
print(name)
print(age)
print(city)
```

```python
# 2. Take two numbers and perform all arithmetic operations.
a = int(input("Enter 1st number: "))
b = int(input("Etner 2nd number: "))
print("Addition", a+b)
print("Subtraction", a-b)
print("Division", a/b)
print("Multiplication", a*b)
print("Modulus (remainder)", a%b)
print("Floor Divison", a//b)
print("Exponentiation (power)", a**b)
```

```python
# 3. Calculate the area of a circle.
r = int(input("Enter radius: "))
pie = 3.14
area_circle = pie*r*r
print("Area of circler:", area_circle)
```

```python
# 4. Convert Celsius to Fahrenheit
celsius = float(input("Enter the Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print("Temperature in fahrenheit: ", fahrenheit)
```

```python
# 5. Calculate simple interest
p = float(input("Enter the principal amount: "))
r = float(input("Rate of interest: "))
t = float(input("Time Period in year: "))
si = (p*r*t)/100
print("Simple Interest is", si)
```

```python
# 6. Swap 2 numbers
a = int(input("Enter a number: "))
b = int(input("Enter b number: "))
temp = a
a = b
b = temp
print("Value of a", a)
print("Value of b", b)
```

```python
# 7. Calculate the average of 5 numbers
a = int(input("Enter 1 number: "))
b = int(input("Enter 2 number: "))
c = int(input("Enter 3 number: "))
d = int(input("Enter 4 number: "))
e = int(input("Enter 5 number: "))
avg_5 = (a+b+c+d+e)/5
print("Average of 5 numbers", avg_5)
```

```python
# Mini Challenge
a = int(input("Enter 1 number: "))
b = int(input("Enter 2 number: "))
c = int(input("Enter 3 number: "))
d = int(input("Enter 4 number: "))
e = int(input("Enter 5 number: "))
total = a+b+c+d+e
avg_5 = total/5
perc = (total/500)*100
print("Total:", total)
print("Average: ", avg_5)
print("Precentage", perc)
```
