```python
class Stack:
  def __init__(self):
    self.st = []

  def push(self,x):
    self.st.append(x)

  def pop(self):
    if len(self.st) == 0:
      return -1

    x = self.st[-1]
    self.st.pop()
    return x

  def top():
    if len(self.st) == 0:
      return -1
    return self.st[-1]

  def size():
    return len(self.st)

stack = Stack()
stack.push(7)
stack.push(8)
stack.push(9)

print(stack.pop())
print(stack.top())
print(stack.size())
```
